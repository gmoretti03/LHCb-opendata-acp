```python
import numpy as np
import matplotlib.pyplot as plt

True_label = 1.0

# Define the loss functions
def mse_loss(p_pred, y_true):
    return (p_pred - y_true)**2

def bce_loss(p_pred, y_true):
    # avoid log(0)
    eps = 1e-15
    p_pred = np.clip(p_pred, eps, 1 - eps)
    return -(y_true * np.log(p_pred) + (1 - y_true) * np.log(1 - p_pred))

# Visualize the "penalty" of the two losses
p_range = np.linspace(0.001, 0.999, 100)
loss_mse = mse_loss(p_range, True_label) # If the true label is 1
loss_bce = bce_loss(p_range, True_label) # If the true label is 1

plt.figure(figsize=(10, 5))
plt.plot(p_range, loss_mse, label='MSE Loss (Quadratic)', lw=2)
plt.plot(p_range, loss_bce, label='BCE Loss (Logarithmic)', lw=2)
plt.title("Loss if True Label is 1 (Signal)")
plt.xlabel("Predicted Probability")
plt.ylabel("Loss Value")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```
